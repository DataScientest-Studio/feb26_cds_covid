```python
def mean_intensity(img):
    arr = np.array(img.convert("L"))
    return arr.mean()

def contrast(img):
    arr = np.array(img.convert("L"))
    return arr.std()
```

```python
records = []
sizes = []
corrupted_files = []
class_counts = {}
mode_counts = Counter()

# Nous allons itérer directement sur les fichiers dans dataset_path
# et déduire la classe du nom du fichier.

# Détecter les classes uniques à partir des noms de fichiers
# En supposant que les noms de fichiers commencent par 'COVID-' ou 'Normal-'

# Utilisation d'un ensemble pour stocker les noms de classes uniques trouvés
unique_classes = set()
for img_path in dataset_path.glob("*.png"):
    filename = img_path.name
    if filename.startswith("COVID-"):
        unique_classes.add("COVID")
    elif filename.startswith("Normal-"):
        unique_classes.add("Normal")

classes = list(unique_classes)
print(f"Classes détectées : {classes}\n")

# Maintenant, itérer sur tous les fichiers PNG dans le répertoire du dataset
for img_path in dataset_path.glob("*.png"):
    filename = img_path.name
    cls = None
    if filename.startswith("COVID-"):
        cls = "COVID"
    elif filename.startswith("Normal-"):
        cls = "Normal"

    if cls is None: # Ignorer les fichiers qui ne correspondent à aucune classe connue
        continue

    if cls not in class_counts:
        class_counts[cls] = 0

    try:
        with Image.open(img_path) as img:

            width, height = img.size
            mode = img.mode

            # features visuelles
            intensity = mean_intensity(img)
            cont = contrast(img)

            sizes.append((width, height))
            mode_counts[mode] += 1

            records.append({
                "class": cls,
                "filename": img_path.name,
                "width": width,
                "height": height,
                "mode": mode,
                "mean_intensity": intensity,
                "contrast": cont,
                "filepath": str(img_path)
            })

            class_counts[cls] += 1

    except Exception:
        corrupted_files.append(str(img_path))
```

```python
from imblearn.over_sampling import RandomOverSampler
import pandas as pd

# Séparation des caractéristiques (X) et la cible (y) de l'ensemble d'entraînement
# Nous incluons 'filepath' dans X_train pour qu'il soit resample et puisse être utilisé par ImageDataGenerator.
# X_train contient: width, height, mean_intensity, contrast, filepath
X_train = train_df.drop(columns=["target", "class", "filename", "mode"])
y_train = train_df["target"]

# Initialisation de RandomOverSampler
# random_state=42 pour la reproductibilité
ros = RandomOverSampler(random_state=42)

# Appliquation de suréchantillonnage
# X_resampled et y_resampled contiendront les échantillons (avec des duplicats) et leurs labels correspondants.
# X_resampled contiendra également la colonne 'filepath'.
X_resampled, y_resampled = ros.fit_resample(X_train, y_train)

print("Distribution des classes après suréchantillonnage (RandomOverSampler) :")
print(y_resampled.value_counts())

print("\nDistribution proportionnelle des classes après suréchantillonnage :")
print(y_resampled.value_counts(normalize=True) * 100)

# Reconstruire train_df_resampled directement à partir de X_resampled et y_resampled.
# X_resampled contient déjà toutes les colonnes de caractéristiques, y compris 'filepath'.
train_df_resampled = X_resampled.copy()
train_df_resampled["target"] = y_resampled.values

print("\nShape du DataFrame d'entraînement après suréchantillonnage:")
print(train_df_resampled.shape)
```

```python
from pathlib import Path

# Dossier contenant les images à prétraiter
INPUT_FOLDER = Path("../covid/test_preprocessing") #to replace at the end with real path containing all images

# Paramètres du preprocessing
IMAGE_SIZE = (224, 224)

COLOR_MODE = "L"          # "L" ou "RGB"

NORMALIZE = True
```

```python
# fonction de prétraitement permettant de modifier les paramètres en fonction du modèle futur
from PIL import Image
import numpy as np


def preprocess_image(
    image_path,
    image_size=IMAGE_SIZE,
    color_mode=COLOR_MODE,
    normalize=NORMALIZE
):
    """
    Charge une image et applique les prétraitements.

    Parameters
    ----------
    image_path : Path
        Chemin de l'image.

    image_size : tuple
        Taille finale.

    color_mode : str
        "L" ou "RGB".

    normalize : bool
        Active ou non la normalisation.

    Returns
    -------
    numpy.ndarray
    """

    with Image.open(image_path) as img:

        img = img.convert(color_mode)

        img = img.resize(image_size)

        img = np.asarray(
            img,
            dtype=np.float32
        )

        if normalize:
            img /= 255.0

    return img
```

```python
import shutil
from pathlib import Path

# Dossier de destination de la copie = input pour la partie preprocessing
test_folder = Path("../covid/test_preprocessing")

if test_folder.exists():
    shutil.rmtree(test_folder)

test_folder.mkdir(parents=True)
```

```python
selected = pd.concat([

    # Une image par mode de couleur
    df.groupby("mode").head(1),

    # Image la plus sombre
    df.nsmallest(1, "mean_intensity"),

    # Image la plus claire
    df.nlargest(1, "mean_intensity"),

    # Contraste minimal
    df.nsmallest(1, "contrast"),

    # Contraste maximal
    df.nlargest(1, "contrast"),

    # Une image COVID
    df[df["class"] == "COVID"].sample(
        n=1,
        random_state=42
    ),

    # Une image NON-COVID
    df[df["class"] != "COVID"].sample(
        n=1,
        random_state=42
    )

])

# Suppression des doublons éventuels
selected = selected.drop_duplicates(subset="filepath")

print(f"{len(selected)} image(s) sélectionnée(s).")

display(
    selected[
        [
            "filename",
            "class",
            "mode",
            "mean_intensity",
            "contrast"
        ]
    ]
)

for _, row in selected.iterrows():

    shutil.copy(
        row["filepath"],
        test_folder / Path(row["filepath"]).name
    )

print(f"\nJeu de test créé dans : {test_folder.resolve()}")
```

```python
#Preprocessing

processed_images = []

supported_extensions = {
    ".png",
    ".jpg",
    ".jpeg",
    ".bmp",
    ".tif",
    ".tiff"
}

for image_path in test_folder.iterdir():

    if image_path.suffix.lower() not in supported_extensions:
        continue

    processed = preprocess_image(image_path)

    processed_images.append(processed)

print(f"{len(processed_images)} image(s) prétraitée(s).")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Définition de l'architecture du modèle
```

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 1 if COLOR_MODE == 'L' else 3)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid') # Sigmoid pour classification binaire
])

model.summary()
```

```python
# Compilation du modèle
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

```python
# Préparation des générateurs de données
```

```python
train_datagen = ImageDataGenerator(rescale=1./255 if not NORMALIZE else None)
                                  # Vous pouvez ajouter des augmentations ici si vous le souhaitez, par exemple:
                                  # rotation_range=20,
                                  # width_shift_range=0.2,
                                  # height_shift_range=0.2,
                                  # shear_range=0.2,
                                  # zoom_range=0.2,
                                  # horizontal_flip=True,
                                  # fill_mode='nearest')

test_datagen = ImageDataGenerator(rescale=1./255 if not NORMALIZE else None)

# Convert 'target' column to string type for ImageDataGenerator
train_df_resampled['target'] = train_df_resampled['target'].astype(str)
test_df['target'] = test_df['target'].astype(str)

# train_df_resampled contient déjà le chemin complet des images dans 'filepath'
# Il est important que le colonne 'target' soit un entier ou un float pour 'class_mode="binary"'
# Assurez-vous que COLOR_MODE est correctement défini ('grayscale' ou 'rgb')

train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_df_resampled,
    x_col='filepath',
    y_col='target',
    target_size=IMAGE_SIZE,
    color_mode='grayscale' if COLOR_MODE == 'L' else 'rgb',
    batch_size=32,
    class_mode='binary',
    shuffle=True,
    seed=42
)

test_generator = test_datagen.flow_from_dataframe(
    dataframe=test_df,
    x_col='filepath',
    y_col='target',
    target_size=IMAGE_SIZE,
    color_mode='grayscale' if COLOR_MODE == 'L' else 'rgb',
    batch_size=32,
    class_mode='binary',
    shuffle=False # Pas besoin de mélanger les données de test
)
```

```python
# Entraînement du modèle
```

```python
history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // train_generator.batch_size,
    epochs=10, # Vous pouvez ajuster le nombre d'époques
    validation_data=test_generator,
    validation_steps=test_generator.samples // test_generator.batch_size
)
```
