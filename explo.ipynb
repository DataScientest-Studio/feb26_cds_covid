```python
#distribution des valeurs 
seuil = 11

for col in df.select_dtypes(include=["object", "category"]).columns:
    nb_modalites = df[col].nunique(dropna=False)

    if nb_modalites < seuil:
        print(f"\n{col} ({nb_modalites} modalités) :")
        print(df[col].unique())
```
