```python
def mean_intensity(img):
    arr = np.array(img.convert("L"))
    return arr.mean()

def contrast(img):
    arr = np.array(img.convert("L"))
    return arr.std()
```

```python
records = []
sizes = []
corrupted_files = []
class_counts = {}
mode_counts = Counter()

# Nous allons itérer directement sur les fichiers dans dataset_path
# et déduire la classe du nom du fichier.

# Détecter les classes uniques à partir des noms de fichiers
# En supposant que les noms de fichiers commencent par 'COVID-' ou 'Normal-'

# Utilisation d'un ensemble pour stocker les noms de classes uniques trouvés
unique_classes = set()
for img_path in dataset_path.glob("*.png"):
    filename = img_path.name
    if filename.startswith("COVID-"):
        unique_classes.add("COVID")
    elif filename.startswith("Normal-"):
        unique_classes.add("Normal")

classes = list(unique_classes)
print(f"Classes détectées : {classes}\n")

# Maintenant, itérer sur tous les fichiers PNG dans le répertoire du dataset
for img_path in dataset_path.glob("*.png"):
    filename = img_path.name
    cls = None
    if filename.startswith("COVID-"):
        cls = "COVID"
    elif filename.startswith("Normal-"):
        cls = "Normal"

    if cls is None: # Ignorer les fichiers qui ne correspondent à aucune classe connue
        continue

    if cls not in class_counts:
        class_counts[cls] = 0

    try:
        with Image.open(img_path) as img:

            width, height = img.size
            mode = img.mode

            # features visuelles
            intensity = mean_intensity(img)
            cont = contrast(img)

            sizes.append((width, height))
            mode_counts[mode] += 1

            records.append({
                "class": cls,
                "filename": img_path.name,
                "width": width,
                "height": height,
                "mode": mode,
                "mean_intensity": intensity,
                "contrast": cont,
                "filepath": str(img_path)
            })

            class_counts[cls] += 1

    except Exception:
        corrupted_files.append(str(img_path))
```

```python
from pathlib import Path

# Dossier contenant les images à prétraiter
INPUT_FOLDER = Path("../covid/test_preprocessing") #to replace at the end with real path containing all images

# Paramètres du preprocessing
IMAGE_SIZE = (224, 224)

COLOR_MODE = "L"          # "L" ou "RGB"

NORMALIZE = True
```

```python
# fonction de prétraitement permettant de modifier les paramètres en fonction du modèle futur
from PIL import Image
import numpy as np


def preprocess_image(
    image_path,
    image_size=IMAGE_SIZE,
    color_mode=COLOR_MODE,
    normalize=NORMALIZE
):
    """
    Charge une image et applique les prétraitements.

    Parameters
    ----------
    image_path : Path
        Chemin de l'image.

    image_size : tuple
        Taille finale.

    color_mode : str
        "L" ou "RGB".

    normalize : bool
        Active ou non la normalisation.

    Returns
    -------
    numpy.ndarray
    """

    with Image.open(image_path) as img:

        img = img.convert(color_mode)

        img = img.resize(image_size)

        img = np.asarray(
            img,
            dtype=np.float32
        )

        if normalize:
            img /= 255.0

    return img
```

```python
selected = pd.concat([

    # Une image par mode de couleur
    df.groupby("mode").head(1),

    # Image la plus sombre
    df.nsmallest(1, "mean_intensity"),

    # Image la plus claire
    df.nlargest(1, "mean_intensity"),

    # Contraste minimal
    df.nsmallest(1, "contrast"),

    # Contraste maximal
    df.nlargest(1, "contrast"),

    # Une image COVID
    df[df["class"] == "COVID"].sample(
        n=1,
        random_state=42
    ),

    # Une image NON-COVID
    df[df["class"] != "COVID"].sample(
        n=1,
        random_state=42
    )

])

# Suppression des doublons éventuels
selected = selected.drop_duplicates(subset="filepath")

print(f"{len(selected)} image(s) sélectionnée(s).")

display(
    selected[
        [
            "filename",
            "class",
            "mode",
            "mean_intensity",
            "contrast"
        ]
    ]
)

for _, row in selected.iterrows():

    shutil.copy(
        row["filepath"],
        test_folder / Path(row["filepath"]).name
    )

print(f"\nJeu de test créé dans : {test_folder.resolve()}")
```

```python
#Preprocessing

processed_images = []

supported_extensions = {
    ".png",
    ".jpg",
    ".jpeg",
    ".bmp",
    ".tif",
    ".tiff"
}

for image_path in test_folder.iterdir():

    if image_path.suffix.lower() not in supported_extensions:
        continue

    processed = preprocess_image(image_path)

    processed_images.append(processed)

print(f"{len(processed_images)} image(s) prétraitée(s).")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau


def create_cnn_model(input_shape, num_classes):
    model = Sequential([
        Conv2D(32, (3, 3), activation='relu', input_shape=input_shape),
        MaxPooling2D((2, 2)),
        Conv2D(64, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Conv2D(128, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Flatten(),
        Dense(128, activation='relu'),
        Dropout(0.5), # Ajout du dropout pour la régularisation.
        Dense(num_classes, activation='sigmoid') # Sigmoïde pour la classification binaire
    ])
    return model

input_channels = 1 if COLOR_MODE == 'L' else 3
input_shape = (*IMAGE_SIZE, input_channels)
num_classes = 1 

# creation du modèle CNN
model = create_cnn_model(input_shape, num_classes)

# Compiler le modèle
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy', # Utilisation de binary_crossentropy pour la classification binaire.
    metrics=['accuracy']
)

model.summary()
```

```python
# Définition del'arrêt anticipé (EarlyStopping) pour éviter le surapprentissage.

early_stopping = EarlyStopping(
    monitor='val_loss', # Surveillence la perte de validation
    patience=10,        # Nombre d'époques sans amélioration au-delà duquel l'entraînement sera arrêté
    restore_best_weights=True # Restauration des poids du modèle à partir de l'époque ayant la meilleure valeur de la grandeur surveillée.
)

# Définition de ReduceLROnPlateau pour ajuster le taux d'apprentissage.
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss', # Surveillence la perte de validation
    factor=0.2,         # Facteur par lequel le taux d'apprentissage sera réduit. new_lr = lr * factor
    patience=5,         # Nombre d'époques sans amélioration au bout duquel le taux d'apprentissage sera réduit
    min_lr=0.00001,     # Borne inférieure du taux d'apprentissage
    verbose=1
)

# Liste des callbacks à utiliser pendant l'entraînement du modèle
callbacks = [early_stopping, reduce_lr]
```

```python
# EVALUATION DU MODÈLE : MATRICE DE CONFUSION
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import numpy as np

#  etiquettes réelles pour le jeu de test
y_true = test_generator.classes

# probabilités prédites par le modèle pour le jeu de test
y_pred_proba = model.predict(test_generator)


y_pred = (y_pred_proba > 0.5).astype(int)


conf_matrix = confusion_matrix(y_true, y_pred)


class_names = list(test_generator.class_indices.keys())

plt.figure(figsize=(8, 6))
sns.heatmap(
    conf_matrix,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names
)
plt.title('Matrice de Confusion')
plt.xlabel('Prédiction')
plt.ylabel('Vraie Valeur')
plt.show()

# Imprimer le rapport de classification pour obtenir des métriques plus détaillées.
print("\nRapport de Classification:")
print(classification_report(y_true, y_pred, target_names=class_names))
```
